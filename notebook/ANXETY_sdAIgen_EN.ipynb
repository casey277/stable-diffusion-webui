```python
import subprocess, time

# Install (idempotent)
!curl -fsSL https://github.com/filebrowser/filebrowser/releases/latest/download/linux-amd64-filebrowser.tar.gz -o /tmp/fb.tar.gz
!tar xzf /tmp/fb.tar.gz -C /usr/local/bin/ && chmod +x /usr/local/bin/filebrowser

# Start on 8090 with a fresh DB
log = open("/tmp/fb.log", "w")
subprocess.Popen(
    ["filebrowser", "--noauth", "-a", "0.0.0.0", "-r", "/content",
     "-p", "8090", "-d", "/tmp/fb.db"],
    stdout=log, stderr=log, start_new_session=True,
)
time.sleep(3)

print("--- FileBrowser log ---")
print(open("/tmp/fb.log").read())

code = subprocess.run(
    ["curl", "-s", "-o", "/dev/null", "-w", "%{http_code}", "http://localhost:8090"],
    capture_output=True, text=True).stdout.strip()
print(f"🔎 localhost:8090 → HTTP {code}")

if code == "200":
    print("✅ FileBrowser is up. Opening tunnel...\n")
    !npx localtunnel --port 8090
else:
    print("❌ Not serving — paste the log above.")
```
